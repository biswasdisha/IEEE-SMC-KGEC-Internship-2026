Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("ggplot")
```

Enter Your Results

```python
results = {

    "Model": [
        "Custom CNN",
        "ResNet50",
        "DenseNet121"
    ],

    "Accuracy (%)": [
        80.13,
        87.82,
        86.86
    ],

    "Precision (%)": [
        76.39,
        85.52,
        83.19
    ],

    "Recall (%)": [
        98.72,
        96.92,
        98.97
    ],

    "AUC (%)": [
        90.53,
        94.81,
        95.73
    ],

    "Loss": [
        0.8393,
        0.4174,
        0.3694
    ]
}

df = pd.DataFrame(results)

df
```

Create Comparison Table

```python
df = pd.DataFrame(results)

df
```

Precision Comparison

```python
plt.figure(figsize=(7,5))

plt.bar(df["Model"], df["Precision (%)"])

plt.title("Precision Comparison")

plt.ylabel("Precision (%)")

plt.show()
```

Recall Comparison

```python
plt.figure(figsize=(7,5))

plt.bar(df["Model"], df["Recall (%)"])

plt.title("Recall Comparison")

plt.ylabel("Recall (%)")

plt.show()
```

AUC Comparison

```python
plt.figure(figsize=(7,5))

plt.bar(df["Model"], df["AUC (%)"])

plt.title("AUC Comparison")

plt.ylabel("AUC (%)")

plt.show()
```

Loss Comparison

```python
plt.figure(figsize=(7,5))

plt.bar(df["Model"], df["Loss"])

plt.title("Loss Comparison")

plt.ylabel("Loss")

plt.show()
```

Overall Ranking

```python
ranking = df.sort_values(
    by="Accuracy (%)",
    ascending=False
)

ranking
```

Best Model

```python
best_model = ranking.iloc[0]

print("="*50)

print("Best Model")

print("="*50)

print(best_model)

print("="*50)
```
